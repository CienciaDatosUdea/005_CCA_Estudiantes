ENTRADA: imágenes originales y niveles de ruido σ

Normalizar las imágenes
Separar datos de entrenamiento y evaluación
Inicializar los parámetros θ del modelo

REPETIR hasta completar el entrenamiento:
    Seleccionar un lote de imágenes originales x
    Seleccionar un nivel de ruido σ

    Generar ruido nuevo: e = σ · Normal(0, I)
    Construir imágenes perturbadas: y = x + e

    Calcular el campo objetivo: S_objetivo = −e / σ²
    Predecir el campo: S_predicho = sθ(y, σ)

    Pérdida = ½ · promedio(||S_predicho − S_objetivo||²)
    Actualizar θ usando el gradiente de la pérdida

Evaluar el modelo con los datos reservados
SALIDA: campo aprendido sθ(y, σ)

```python
# Ruuido gaussian
Sobjetivo = -e/sigma**2
plt.imshow(Sobjetivo.reshape(N, N))
```

```python
xp = x + e*sigma
plt.imshow(xp.reshape(8,8))
```

```python
A = 0.0
B = np.zeros(len(xp))
alphaA = 0.5
alphaB = 5.0
rng = np.random.default_rng(42)

for i in range(0, 50000):
  e = rng.normal(0, sigma, size = x.shape)
  xp = x + e
  Z = A*xp + B
  Sobjetivo = -e/sigma**2
  Spredicho = Z.copy()
  error = Spredicho - Sobjetivo
  loss = 0.5*np.mean(error**2)
  gradA = np.mean(error*xp)
  gradB = error/x.size
  A = A-alphaA*gradA
  B = B-alphaB*gradB

print(loss)
```

```python
Y = xp + sigma**2*Spredicho
plt.imshow(Y.reshape(8,8))
```

# Langevine

ENTRADA: campo aprendido, niveles σ de mayor a menor,
         número de pasos K y constante pequeña c

Inicializar desde ruido:
    Y = μ + σ_max · Normal(0, I)
    μ es la media de las imágenes de entrenamiento

PARA cada nivel σ, desde σ_max hasta σ_min:
    Elegir el paso: h = c · σ²

    REPETIR K veces:
        Calcular el campo: S = sθ(Y, σ)
        Generar ruido nuevo: Z = Normal(0, I)

        Actualizar mediante Langevin:
            Y = Y + h · S + √(2h) · Z

Opcionalmente aplicar la recuperación final:
    Y = Y + σ_min² · sθ(Y, σ_min)

SALIDA: imagen generada

```python
rng_generacion = np.random.default_rng(123)

# Inicialización independiente de la imagen original.
Y = rng_generacion.normal(0, 1, size=x.shape)
Y_inicial = Y.copy()

dt = 0.001  # Para sigma = 0.1 y A cercano a -100.
n_pasos = 1000

# Aquí entra Langevin.
for paso in range(n_pasos):

    S = A * Y + B
    Z = rng_generacion.normal(size=Y.shape)
    Y = Y + dt * S + np.sqrt(2 * dt) * Z

# Recuperación final: es una operación distinta.
Y_recuperada = Y + sigma**2 * (A * Y + B)

fig, ejes = plt.subplots(1, 3, figsize=(9, 3))

comparaciones = [
    (Y_inicial, "Inicialización aleatoria"),
    (Y, "Después de Langevin"),
    (Y_recuperada, "Recuperación final")
]

for eje, (imagen, titulo) in zip(ejes, comparaciones):
    eje.imshow(
        np.clip(imagen.reshape(N, N), 0, 1),
        cmap="gray", vmin=0, vmax=1
    )
    eje.set_title(titulo)
    eje.axis("off")

plt.tight_layout()
plt.show()
```

```python
n_pasos = 2000
historia = []
for paso in range(n_pasos):
    # Seleccionar un grupo de imágenes
    indices = rng.integers(0, len(datos), size=32)
    X = datos[indices]
    # Agregar ruido nuevo
    epsilon = rng.normal(0, sigma, size=X.shape).astype(np.float32)
    Y = X + epsilon
    # Score objetivo y score predicho
    S_objetivo = -epsilon / sigma**2

    S_predicho = a * Y + B



    error = S_predicho - S_objetivo
    loss = 0.5 * np.mean(error**2)
    # Gradientes de esa pérdida
    grad_a = np.mean(error * Y)
    grad_B = np.mean(error, axis=0) / D

    # Actualizar los parámetros
    a -= lr_a * grad_a
    B -= lr_B * grad_B

    historia.append(loss)

    if (paso + 1) % 200 == 0:
        print(f"Paso {paso + 1}: pérdida = {loss:.4f}")

print("a aprendido:", a)

plt.plot(historia)
plt.xlabel("Paso")
plt.ylabel("Pérdida")
plt.show()
```

```python
import json
from urllib.request import urlopen
from io import BytesIO
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps

N = 100  # Puedes empezar con 20 para descargar todavía menos.
TAMANO = 64

carpeta = Path("flores_64")
carpeta.mkdir(exist_ok=True)

url = (
    "https://datasets-server.huggingface.co/rows?"
    "dataset=nkirschi/oxford-flowers&config=default&split=train"
    f"&offset=0&length={N}"
)

# Consultar únicamente las imágenes solicitadas.
with urlopen(url, timeout=60) as respuesta:
    filas = json.load(respuesta)["rows"]

imagenes = []

for i, fila in enumerate(filas):
    archivo = carpeta / f"flor_{i:03d}.png"

    if not archivo.exists():
        with urlopen(fila["row"]["image"]["src"], timeout=60) as respuesta:
            imagen = Image.open(BytesIO(respuesta.read())).convert("RGB")

        imagen = ImageOps.fit(imagen, (TAMANO, TAMANO))
        imagen.save(archivo)

    with Image.open(archivo) as imagen:
        imagenes.append(np.array(imagen))

# Valores entre 0 y 1, listos para trabajar después con Keras.
imagenes = np.array(imagenes, dtype="float32") / 255.0
print("Forma:", imagenes.shape)
```

```python
rng = np.random.default_rng(42)

# 1. Barajar y separar las imágenes.
indices = rng.permutation(len(imagenes))
corte = int(0.8 * len(imagenes))

entrenamiento = imagenes[indices[:corte]]
test = imagenes[indices[corte:]]

print("Entrenamiento:", entrenamiento.shape)
print("Test:", test.shape)

# 2. Perturbar todos los píxeles y canales.
sigma = 0.20  # Intensidad del ruido; las imágenes están entre 0 y 1.

def perturbar(X, sigma):
    ruido = rng.normal(size=X.shape).astype("float32")
    return X + sigma * ruido, ruido

entrenamiento_perturbado, ruido_train = perturbar(entrenamiento, sigma)
test_perturbado, ruido_test = perturbar(test, sigma)

# 3. Comparar una imagen antes y después.
i = 0

fig, ejes = plt.subplots(1, 2, figsize=(6, 3))

ejes[0].imshow(entrenamiento[i])
ejes[0].set_title("Original")

# Recortamos a [0, 1] únicamente para visualizar.
ejes[1].imshow(np.clip(entrenamiento_perturbado[i], 0, 1))
ejes[1].set_title(f"Perturbada: σ = {sigma}")

for eje in ejes:
    eje.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

import keras
from keras import layers

keras.utils.set_random_seed(42)

modelo = keras.Sequential([
    keras.Input(shape=entrenamiento.shape[1:]),

    layers.Conv2D(
        32, kernel_size=3, padding="same", activation="relu"
    ),
    layers.Conv2D(
        32, kernel_size=3, padding="same", activation="relu"
    ),

    # Predecir el ruido de los tres canales RGB.
    layers.Conv2D(3, kernel_size=3, padding="same")
])

modelo.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="mse"
)

modelo.summary()

print("Entrada:", modelo.input_shape)
print("Salida:", modelo.output_shape)
```

```python
sigma = 0.15
n_epocas = 20

perdidas_train = []
perdidas_test = []

# El test mantiene el mismo ruido durante toda la evaluación.
test_perturbado, ruido_test = perturbar(test, sigma)

for epoca in range(n_epocas):
    # Entrenamiento con ruido nuevo.
    entradas, ruido_real = perturbar(entrenamiento, sigma)

    historia = modelo.fit(
        entradas,
        ruido_real,
        epochs=1,
        batch_size=8,
        shuffle=True,
        verbose=0
    )

    mse_train = historia.history["loss"][0]

    # Evaluación: los pesos permanecen fijos.
    mse_test = modelo.evaluate(
        test_perturbado,
        ruido_test,
        batch_size=8,
        verbose=0
    )

    perdidas_train.append(mse_train)
    perdidas_test.append(mse_test)

    print(
        f"Época {epoca + 1:02d}/{n_epocas} | "
        f"Train: {mse_train:.4f} | Test: {mse_test:.4f}"
    )

# Graficar ambas curvas.
epocas = range(1, n_epocas + 1)
```

```python
import torch

n_imagenes = 9
n_pasos = 5000
rng_generacion = np.random.default_rng(123)

# Paso pequeño de Langevin.
h = np.float32(0.005 * sigma**2)
amplitud = np.float32(np.sqrt(2 * h))

# Inicialización independiente de las flores originales.
forma = tuple(modelo.input_shape[1:])
generadas = rng_generacion.uniform(
    0, 1, size=(n_imagenes, *forma)
).astype("float32")

# Guardar la evolución de la primera imagen.
evolucion = {0: generadas[0].copy()}
guardar_en = {50, 100, 250, n_pasos}

# Inferencia con tu backend Torch: los pesos permanecen fijos.
with torch.no_grad():
    for paso in range(1, n_pasos + 1):

        # La red predice epsilon, no sigma * epsilon.
        ruido_predicho = keras.ops.convert_to_numpy(
            modelo(generadas, training=False)
        )

        score = -ruido_predicho / sigma

        # Ruido nuevo para la dinámica de Langevin.
        z = rng_generacion.normal(
            size=generadas.shape
        ).astype("float32")

        generadas = generadas + h * score + amplitud * z

        # Mantener los valores sin recortar durante la dinámica.
        if paso in guardar_en:
            evolucion[paso] = generadas[0].copy()

        if paso % 100 == 0:
            if not np.isfinite(generadas).all():
                raise FloatingPointError(
                    "La dinámica diverge. Reduce el valor de h."
                )
            print(f"Paso {paso}/{n_pasos}")

```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


def agregar_sigma(X, sigma, sigma_max):
    """
    Agrega un canal constante con sigma normalizado.
    X: (..., alto, ancho, canales)
    """
    X = X.astype(np.float32)
    canal_sigma = np.full(
        X.shape[:-1] + (1,),
        sigma / sigma_max,
        dtype=np.float32
    )
    return np.concatenate([X, canal_sigma], axis=-1)


def bloque_conv(x, filtros):
    x = layers.Conv2D(
        filtros, 3, padding="same", kernel_initializer="he_normal"
    )(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)

    x = layers.Conv2D(
        filtros, 3, padding="same", kernel_initializer="he_normal"
    )(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)
    return x


def construir_unet(forma_entrada, canales_salida, filtros_base=32, profundidad=3):
    """
    U-Net estándar con capas de Keras.
    Entrada: imagen + canal sigma.
    Salida: campo score con los canales de la imagen.
    """
    entrada = layers.Input(shape=forma_entrada)

    x = entrada
    saltos = []

    # Encoder
    for i in range(profundidad):
        filtros = filtros_base * (2 ** i)
        x = bloque_conv(x, filtros)
        saltos.append(x)
        x = layers.MaxPooling2D((2, 2))(x)

    # Bottleneck
    filtros = filtros_base * (2 ** profundidad)
    x = bloque_conv(x, filtros)

    # Decoder
    for i in reversed(range(profundidad)):
        filtros = filtros_base * (2 ** i)
        x = layers.Conv2DTranspose(
            filtros, 2, strides=2, padding="same"
        )(x)
        x = layers.Concatenate()([x, saltos[i]])
        x = bloque_conv(x, filtros)

    # Score no acotado: sin sigmoid ni tanh
    salida = layers.Conv2D(
        canales_salida,
        1,
        padding="same",
        activation=None,
        kernel_initializer="zeros",
        bias_initializer="zeros"
    )(x)

    return keras.Model(entrada, salida, name="unet_score")
```

```python
X_entrenamiento = imagenes.astype(np.float32) * 2 - 1

corte = int(0.8 * len(X_entrenamiento))
datos = X_entrenamiento[:corte].astype(np.float32)
evaluacion = X_entrenamiento[corte:].astype(np.float32)

alto, ancho, canales = X_entrenamiento.shape[1:]
forma_entrada = (alto, ancho, canales + 1)

# Varios niveles de ruido para que el modelo sirva también en Langevin
sigmas = np.array([1.0, 0.5, 0.3, 0.2, 0.1, 0.05], dtype=np.float32)
sigma_max = float(sigmas.max())
sigma_eval = 0.3

keras.backend.clear_session()

profundidad = 3 if alto <= 32 else 4

modelo_score = construir_unet(
    forma_entrada,
    canales_salida=canales,
    filtros_base=32,
    profundidad=profundidad
)

modelo_score.summary()

def perdida_score(y_true, y_pred):
    return 0.5 * tf.reduce_mean((y_pred - y_true) ** 2)

modelo_score.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3, clipnorm=1.0),
    loss=perdida_score
)
```

```python
import time

optimizador = tf.keras.optimizers.Adam(learning_rate=3e-3, clipnorm=1.0)

# Firma fija -> tf.function compila una sola vez
@tf.function(reduce_retracing=True)
def paso_entrenamiento(Y_sigma, S_objetivo):
    with tf.GradientTape() as tape:
        S_pred = modelo_score(Y_sigma, training=True)
        loss = 0.5 * tf.reduce_mean((S_pred - S_objetivo) ** 2)
    grads = tape.gradient(loss, modelo_score.trainable_variables)
    optimizador.apply_gradients(zip(grads, modelo_score.trainable_variables))
    return loss

# Warm-up (compilación)
dummy_x = tf.zeros((64,) + forma_entrada, dtype=tf.float32)
dummy_y = tf.zeros((64,) + forma_entrada[:-1] + (canales,), dtype=tf.float32)
_ = paso_entrenamiento(dummy_x, dummy_y)
print("Compilado. Empezando a entrenar...")

rng = np.random.default_rng(42)
n_pasos = 800
tam_lote = 64
historia = []

t0 = time.time()

for paso in range(n_pasos):

    indices = rng.integers(0, len(datos), size=tam_lote)
    X = datos[indices]

    sigma_lote = rng.choice(sigmas, size=tam_lote).astype(np.float32)
    sigma_b = sigma_lote[:, None, None, None]

    epsilon = rng.normal(size=X.shape).astype(np.float32) * sigma_b
    Y = X + epsilon
    S_objetivo = -epsilon / sigma_b**2

    sigma_canal = np.broadcast_to(
        sigma_b / sigma_max, Y.shape[:-1] + (1,)
    ).astype(np.float32)

    Y_sigma = np.concatenate([Y, sigma_canal], axis=-1)

    loss = paso_entrenamiento(
        tf.convert_to_tensor(Y_sigma,  dtype=tf.float32),
        tf.convert_to_tensor(S_objetivo, dtype=tf.float32),
    )
    historia.append(float(loss.numpy()))

    if paso == 0 or (paso + 1) % 100 == 0:
        t = time.time() - t0
        seg_por_paso = t / (paso + 1)
        eta = seg_por_paso * (n_pasos - paso - 1)
        print(f"Paso {paso+1:4d} | loss {float(loss):.4f} | "
              f"{seg_por_paso:.2f}s/paso | ETA {eta/60:.1f} min")

plt.plot(historia)
plt.xlabel("Paso"); plt.ylabel("Pérdida"); plt.show()
```

```python
def predecir_score(modelo_score, Y, sigma, sigma_max):
    entrada = agregar_sigma(Y, sigma, sigma_max)

    un_solo = (entrada.ndim == 3)
    if un_solo:
        entrada = entrada[None, ...]          # (1, alto, ancho, C+1)

    salida = modelo_score(entrada, training=False)
    salida = np.asarray(salida, dtype=np.float32)

    if un_solo:
        salida = salida[0]                    # (alto, ancho, canales)

    return salida


def recuperar_con_unet(X, sigma, modelo_score, sigma_max, rng):
    X = X.astype(np.float32)

    epsilon = rng.normal(
        0, sigma, size=X.shape
    ).astype(np.float32)

    Y = X + epsilon

    S_predicho = predecir_score(
        modelo_score, Y, sigma, sigma_max
    )

    X_recuperada = Y + sigma**2 * S_predicho

    return Y, X_recuperada, S_predicho


rng_eval = np.random.default_rng(123)

X = evaluacion[4]
Y, X_recuperada, S_predicho = recuperar_con_unet(
    X, sigma_eval, modelo_score, sigma_max, rng_eval
)

fig, axes = plt.subplots(1, 3, figsize=(9, 3))

imagenes_mostrar = [X, Y, X_recuperada]
titulos = ["Original", "Con ruido", "Recuperada: U-Net"]

for ax, imagen, titulo in zip(axes, imagenes_mostrar, titulos):
    ax.imshow(np.clip((imagen + 1) / 2, 0, 1))
    ax.set_title(titulo)
    ax.axis("off")

plt.tight_layout()
plt.show()

print("MSE con ruido:", np.mean((Y - X)**2))
print("MSE recuperada:", np.mean((X_recuperada - X)**2))
```

```python
def generar_con_unet(
    modelo_score,
    forma,
    sigmas,
    n_muestras=4,
    K=100,
    c=0.1,
    rng=None
):
    if rng is None:
        rng = np.random.default_rng(123)

    sigma_max = float(sigmas.max())
    sigma_min = float(sigmas.min())

    # Inicialización desde ruido puro
    Y = rng.normal(
        0, sigma_max, size=(n_muestras,) + forma
    ).astype(np.float32)

    # De sigma grande a sigma pequeño
    for sigma in sigmas:
        h = c * float(sigma) ** 2

        for _ in range(K):
            S = predecir_score(
                modelo_score, Y, sigma, sigma_max
            )
            Z = rng.normal(size=Y.shape).astype(np.float32)
            Y = Y + h * S + np.sqrt(2 * h) * Z

    # Recuperación final opcional
    S = predecir_score(
        modelo_score, Y, sigma_min, sigma_max
    )
    Y = Y + sigma_min**2 * S

    return Y


rng_gen = np.random.default_rng(123)

Y_generadas = generar_con_unet(
    modelo_score,
    forma=(alto, ancho, canales),
    sigmas=sigmas,
    n_muestras=4,
    K=100,
    c=0.1,
    rng=rng_gen
)

fig, axes = plt.subplots(1, 4, figsize=(10, 3))
for i, ax in enumerate(axes):
    ax.imshow(np.clip((Y_generadas[i] + 1) / 2, 0, 1))
    ax.set_title(f"Muestra {i + 1}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

# Cierre: U-Net como modelo de score

## Qué se logró

**Recuperación (denoising):** la U-Net aprende a predecir el score `sθ(y, σ) ≈ -e/σ²` y la regla `x = y + σ²·sθ(y, σ)` reconstruye la imagen original con mucha mayor fidelidad que el modelo lineal `a·Y + B`. La primera figura lo confirma: el ruido gaussiano se elimina casi por completo y los bordes del girasol se preservan.

**Generación (Langevin):** el proceso `Y ← Y + h·sθ(Y, σ) + √(2h)·Z` recorre los niveles de σ, pero con la configuración actual (U-Net pequeña, pocos pasos de entrenamiento, `K` bajo) las muestras convergen a patrones de alta frecuencia en lugar de imágenes de flores. La segunda figura muestra este límite.

## Por qué la generación no converge

1. **Capacidad del modelo:** `filtros_base=16, prof=3` (~600k parámetros) es suficiente para denoising puntual pero no para modelar la distribución completa de imágenes en `64×64`.
2. **Entrenamiento corto:** 800 pasos con lotes de 16 no alcanzan a cubrir los seis niveles de σ con calidad.
3. **Parámetros de Langevin:** `K=50, c=0.1` son un compromiso por la RAM; con una GPU y más pasos, el muestreo se estabiliza.

## Comparación con la versión lineal

| Aspecto | Score lineal `a·Y + B` | U-Net `sθ(y, σ)` |
|---|---|---|
| Recuperación | Buena, pero con un único `σ` fijo | Mejor, y generaliza a varios `σ` |
| Generación | Colapsa a la media | Captura estructura local, pero no global |
| Costo | Segundos | Minutos (CPU) |

La U-Net supera al modelo lineal en la tarea para la que fue entrenada (denoising) y deja abierta la generación como una extensión natural si se dispone de más cómputo.
