Existen otro tipo de filtros que pueden ser empleados.

- [Sobel filter](https://en.wikipedia.org/wiki/Sobel_operator)
  ```
  filter = array([[ 1,  0, -1],
       [ 2,  0, -2],
       [ 1,  0, -1]])
```
- Scharr Filter
```
filter = array([[ 3,  0, -3],
       [ 10,  0, -10],
       [ 3,  0, -3]])
```

¿Cómo elegir los parametros del filtro para detectar otro tipo de bordes?
```
filter = array([[ w1,  w2, w3],
     [ w4,  w5, w6],
     [ w7,  w8, w9]])
```

**Respuesta.** En una red convolucional los pesos $w_1,\dots,w_9$ **no se eligen a mano**: se inicializan al azar y se aprenden con *backpropagation*, minimizando la función de pérdida igual que cualquier otro peso de la red. Los filtros de Sobel o Scharr son casos particulares diseñados a mano para detectar bordes; la red puede llegar a filtros parecidos (o a otros) si le sirven para la tarea. Para detectar un borde con otra orientación, por ejemplo diagonal, se puede rotar el filtro: $\begin{bmatrix}0&1&1\\-1&0&1\\-1&-1&0\end{bmatrix}$.


# Padding

El padding permite que la imagen resultante, sea del mismo tamaño que la original. Dada una matrix $n \times n$ asociada a una imagen y aplicada un filtro de rango $f\times f$ la operación después de la convolución es de rango $(n-f+1) \times (n-f+1)$.

Para garantizar que se tenga una imagen con el mismo tamaño de la original se aplica el padding



Para un padding de 1,tenemos entonces la siguientes dimensiones:

###  input = (n)x(n)

## 0. ¿Qué es Keras?

**Keras** es una API de alto nivel para construir y entrenar redes neuronales. Desde la versión 3, Keras es *multi-backend*: el mismo código (`keras.layers`, `keras.Model`, `.compile()`, `.fit()`, ...) puede ejecutarse sobre **TensorFlow**, **JAX** o **PyTorch** como motor de cómputo. El backend se elige con la variable de entorno `KERAS_BACKEND` **antes** de importar `keras`; el resto del notebook no depende de cuál se use.

Comparado con lo que se hace  "a mano" en PyTorch, Keras encapsula el ciclo de entrenamiento:

| PyTorch (manual)                                              | Keras (equivalente)                    |
|---|---|
| Definir `Dataset` + `DataLoader`                               | Pasar arreglos de NumPy directamente a `.fit()` |
| Escribir el bucle `for epoch ... for batch ...`                | `model.fit(X, y, epochs=..., batch_size=...)` |
| `loss.backward()`, `optimizer.step()`, `optimizer.zero_grad()` | Ocurre internamente en `.fit()`        |
| Guardar el mejor modelo con un `if val_loss < best_val_loss`   | Callback `keras.callbacks.ModelCheckpoint` |
| `model.eval()` + `torch.no_grad()` + bucle manual              | `model.evaluate(X, y)` / `model.predict(X)` |

Un modelo en Keras se arma apilando **capas** (`keras.layers`) dentro de un `keras.Sequential` (o la API funcional para arquitecturas más complejas), se **compila** indicando optimizador, función de pérdida y métricas, y se **entrena** con `.fit()`.

````python
import os
os.environ.setdefault("KERAS_BACKEND", "jax")  # TensorFlow, JAX o PyTorch: el código de Keras no cambia

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import keras
from keras import layers

print("Keras:", keras.__version__)
print("Backend:", keras.backend.backend())
````

## 1. Datos: configuraciones de espines del modelo de Ising

Reutilizamos el dataset generado en `07_Lab_NN_IsingV2` (ver `07_bDatos_Ising.ipynb` para el detalle de cómo se generó). Cada configuración es una matriz $L\times L$ de espines $\pm 1$, etiquetada según su temperatura respecto a $T_c$:

| | fase **ordenada** (`y = 0`) | fase **desordenada** (`y = 1`) |
|---|---|---|
| temperatura | $T < T_c$ | $T > T_c$ |
| aspecto | un dominio grande domina la red | "ruido", sin dominios |

`ising_dataset.npz` **excluye** una banda alrededor de $T_c$ a propósito, para que las etiquetas sean inequívocas durante el entrenamiento.

````python
datos = np.load("Datos/ising_dataset.npz", allow_pickle=False)

X = datos["X"]
y = datos["y"]
T = datos["T"]
m = datos["m"]
realizacion = datos["realizacion"]
Tc_onsager = float(datos["Tc"])
L = int(datos["L"])

print("X:", X.shape, X.dtype)
print("y:", y.shape, y.dtype)
print("Temperaturas:", np.unique(T))
print("Realizaciones:", np.unique(realizacion))
print(f"L = {L}   Tc (Onsager) = {Tc_onsager:.4f}")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(3, 3))

idx_ordenada = np.where(y == 0)[0][0]
idx_desordenada = np.where(y == 1)[0][0]

for ax, idx, titulo in zip(
    axes,
    [idx_ordenada, idx_desordenada],
    [f"Ordenada (y=0)\nT = {T[idx_ordenada]:.2f}", f"Desordenada (y=1)\nT = {T[idx_desordenada]:.2f}"],
):
    ax.imshow(X[idx], cmap="gray", vmin=-1, vmax=1)
    ax.set_title(titulo)
    ax.set_xticks([]); ax.set_yticks([])

plt.tight_layout()
plt.show()
````

## 2. Preparación de los datos para Keras

Igual que en el notebook de PyTorch, usamos el campo `realizacion` (cadena de Monte Carlo independiente) para separar train/val/test, evitando que configuraciones muy correlacionadas de una misma cadena aparezcan en más de un conjunto:

- `realizacion` 0 y 1 → entrenamiento
- `realizacion` 2 → validación
- `realizacion` 3 → test

A diferencia del MLP (que aplanaba cada configuración a un vector de 256 valores), una CNN espera imágenes con forma `(alto, ancho, canales)`. Como tenemos un solo canal (el espín), reformamos cada configuración a `(L, L, 1)`. No hace falta convertir nada a tensores de PyTorch ni armar `DataLoader`s: Keras trabaja directamente con arreglos de NumPy.

````python
mask_train = np.isin(realizacion, [0, 1])
mask_val = realizacion == 2
mask_test = realizacion == 3

def preparar(mask):
    X_ = X[mask].astype("float32").reshape(-1, L, L, 1)
    y_ = y[mask].astype("float32")
    T_ = T[mask]
    return X_, y_, T_

X_train, y_train, T_train = preparar(mask_train)
X_val, y_val, T_val = preparar(mask_val)
X_test, y_test, T_test = preparar(mask_test)

print("Train:", X_train.shape, y_train.shape)
print("Val:  ", X_val.shape, y_val.shape)
print("Test: ", X_test.shape, y_test.shape)
````

## 3. Arquitectura: red convolucional simple

Una CNN es una elección natural para este problema: la fase de una configuración es una propiedad **estadística y global** (¿hay dominios grandes o no?), no depende de *dónde* aparezcan los dominios. Los filtros convolucionales detectan patrones locales (espines alineados con sus vecinos) **sin importar la posición**, exactamente la invariancia que necesitamos.

Arquitectura (deliberadamente simple):

- `Conv2D(8, 3x3) + ReLU` — detecta patrones locales de alineación.
- `MaxPooling2D(2x2)` — reduce la resolución espacial.
- `Conv2D(16, 3x3) + ReLU` — combina patrones locales en patrones un poco más grandes.
- `GlobalAveragePooling2D` — en vez de aplanar (`Flatten`) y usar una capa densa grande, promediamos cada mapa de activación sobre todo el espacio. Esto produce un vector de tamaño fijo **sin importar el tamaño $L\times L$ de la entrada**: la misma red podrá evaluarse más adelante sobre configuraciones de otro tamaño de red, algo que el MLP (con su entrada aplanada de tamaño fijo) no puede hacer.
- `Dense(16) + ReLU` y `Dense(1) + sigmoide` — combinan las características globales en una probabilidad de fase desordenada.

Para permitir tamaños de entrada variables declaramos `Input(shape=(None, None, 1))`.

````python
keras.utils.set_random_seed(42)

model = keras.Sequential(
    [
        keras.Input(shape=(None, None, 1)),
        layers.Conv2D(8, 3, padding="same", activation="relu"),
        layers.MaxPooling2D(2),
        layers.Conv2D(16, 3, padding="same", activation="relu"),
        layers.GlobalAveragePooling2D(),
        layers.Dense(16, activation="relu"),
        layers.Dense(1, activation="sigmoid"),
    ],
    name="ising_cnn",
)

model.summary()
````

## 4. Compilación y entrenamiento

`compile()` fija el optimizador, la función de pérdida y las métricas a reportar. Como la última capa ya tiene activación sigmoide (produce una probabilidad), usamos `binary_crossentropy` directamente (en PyTorch usamos `BCEWithLogitsLoss` porque la red devolvía *logits*, no probabilidades).

El callback `ModelCheckpoint` guarda automáticamente los pesos con menor `val_loss`, reemplazando el bloque manual `if val_loss < best_val_loss: ...` del notebook de PyTorch.

````python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

checkpoint = keras.callbacks.ModelCheckpoint(
    "best_ising_cnn.keras",
    monitor="val_loss",
    save_best_only=True,
)

historia = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=20,
    batch_size=32,
    callbacks=[checkpoint],
    verbose=2,
)
````

````python
epochs_range = range(1, len(historia.history["loss"]) + 1)

plt.figure(figsize=(7, 5))
plt.plot(epochs_range, historia.history["loss"], label="Train")
plt.plot(epochs_range, historia.history["val_loss"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Binary crossentropy")
plt.title("Curva de aprendizaje - Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

````python
plt.figure(figsize=(7, 5))
plt.plot(epochs_range, historia.history["accuracy"], label="Train")
plt.plot(epochs_range, historia.history["val_accuracy"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Curva de aprendizaje - Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.ylim(0.9, 1.005)
plt.show()
````

## 5. Evaluación en el conjunto de test

````python
model = keras.models.load_model("best_ising_cnn.keras")

test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Test loss:     {test_loss:.5f}")
print(f"Test accuracy: {test_acc:.4f}")
````

````python
probs_test = model.predict(X_test, verbose=0).ravel()

unique_T_test = np.unique(T_test)
mean_p, std_p = [], []
for temp in unique_T_test:
    mask = T_test == temp
    mean_p.append(probs_test[mask].mean())
    std_p.append(probs_test[mask].std())
mean_p, std_p = np.array(mean_p), np.array(std_p)

plt.figure(figsize=(7, 5))
plt.errorbar(unique_T_test, mean_p, yerr=std_p, fmt="o-", capsize=4)
plt.axhline(0.5, linestyle="--")
plt.xlabel("Temperatura")
plt.ylabel(r"$\langle P(\mathrm{desordenada} \mid X)\rangle$")
plt.title("Probabilidad de fase desordenada — conjunto de test")
plt.grid(alpha=0.3)
plt.show()
````

## 6. Barrido cerca de la temperatura crítica

`ising_dataset.npz` deja fuera a propósito una banda alrededor de $T_c$. `temperatura_critica.npz` es el conjunto contrario: un barrido fino que **sí incluye** la zona crítica, con varias realizaciones independientes. Aquí no hay etiqueta de entrenamiento confiable — la idea es que, si la red aprendió a reconocer fases, su probabilidad promedio $\langle P(\mathrm{desordenada}\mid T)\rangle$ pase de 0 a 1 alrededor de la verdadera $T_c$, y podamos **estimar $T_c$ como el punto donde esa curva cruza 0.5**.

````python
datos_tc = np.load("Datos/temperatura_critica.npz", allow_pickle=False)

X_tc = datos_tc["X"].astype("float32").reshape(-1, L, L, 1)
T_tc = datos_tc["T"]
realizacion_tc = datos_tc["realizacion"]

print("X_tc:", X_tc.shape)
print("Temperaturas:", np.unique(T_tc))
print("Realizaciones:", np.unique(realizacion_tc))
````

````python
probs_tc = model.predict(X_tc, verbose=0).ravel()
pred_tc = (probs_tc >= 0.5).astype(int)

df = pd.DataFrame({
    "T": T_tc,
    "realizacion": realizacion_tc,
    "probabilidad": probs_tc,
    "prediccion": pred_tc,
})

resumen = (
    df.groupby(["realizacion", "T"])
      .agg(prob_media=("probabilidad", "mean"), prob_std=("probabilidad", "std"), N=("prediccion", "size"))
      .reset_index()
)

resumen.head(10)
````

````python
plt.figure(figsize=(8, 5))

for r in np.unique(realizacion_tc):
    datos_r = resumen[resumen["realizacion"] == r]
    plt.plot(datos_r["T"], datos_r["prob_media"], "o-", label=f"Realización {r}")

plt.axhline(0.5, linestyle="--", label=r"$P=0.5$")
plt.axvline(Tc_onsager, linestyle=":", color="k", label=fr"$T_c$ Onsager = {Tc_onsager:.3f}")

plt.xlabel("Temperatura")
plt.ylabel(r"$\langle P(\mathrm{desordenada})\rangle$")
plt.title("Clasificación de la CNN cerca de la transición de fase")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

````python
Tc_estimadas = []

for r in np.unique(realizacion_tc):
    datos_r = resumen[resumen["realizacion"] == r].sort_values("T")
    temps = datos_r["T"].to_numpy()
    probs = datos_r["prob_media"].to_numpy()

    for i in range(len(temps) - 1):
        p1, p2 = probs[i], probs[i + 1]
        if p1 <= 0.5 <= p2:
            T1, T2 = temps[i], temps[i + 1]
            Tc_r = T1 + (0.5 - p1) * (T2 - T1) / (p2 - p1)  # interpolación lineal
            Tc_estimadas.append(Tc_r)
            print(f"Realización {r}: Tc_CNN = {Tc_r:.4f}")
            break

Tc_estimadas = np.array(Tc_estimadas)
Tc_mean = Tc_estimadas.mean()
Tc_std = Tc_estimadas.std(ddof=1)

print()
print(f"Tc_CNN promedio = {Tc_mean:.4f} +/- {Tc_std:.4f}")
print(f"Tc Onsager      = {Tc_onsager:.4f}")
````

````python
plt.figure(figsize=(8, 5))

for r in np.unique(realizacion_tc):
    datos_r = resumen[resumen["realizacion"] == r]
    plt.plot(datos_r["T"], datos_r["prob_media"], "o-", label=f"Realización {r}")

plt.axhline(0.5, linestyle="--", label=r"$P=0.5$")
plt.axvline(Tc_mean, linestyle="--", color="C3", label=fr"$T_c^{{CNN}} = {Tc_mean:.3f}$")
plt.axvline(Tc_onsager, linestyle=":", color="k", label=fr"$T_c$ Onsager = {Tc_onsager:.3f}")

plt.xlabel("Temperatura")
plt.ylabel(r"$\langle P(\mathrm{desordenada})\rangle$")
plt.title("Estimación de la temperatura crítica con la CNN")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

## 7. Bonus: generalización a otro tamaño de red ($L=20$)

`GlobalAveragePooling2D` hace que la red no tenga ningún peso atado a un tamaño de entrada específico. Podemos entonces tomar exactamente el mismo modelo, **sin reentrenar**, y evaluarlo sobre `temperatura_critica20.npz` (configuraciones $20\times 20$, con la misma $T_c$ física). Con un MLP de entrada aplanada esto sería imposible sin cambiar la arquitectura.

````python
# Tc estimada para L = 20 (misma CNN, sin reentrenar), por realización
resumen20_r = (df20.groupby(["realizacion", "T"])["probabilidad"].mean().reset_index())
Tc20 = []
for r in np.unique(realizacion_tc20):
    d = resumen20_r[resumen20_r["realizacion"] == r].sort_values("T")
    t, p = d["T"].to_numpy(), d["probabilidad"].to_numpy()
    i = np.where((p[:-1] <= 0.5) & (p[1:] >= 0.5))[0][0]
    Tc20.append(t[i] + (0.5 - p[i]) * (t[i+1] - t[i]) / (p[i+1] - p[i]))
    print(f"Realización {r}: Tc_CNN(L=20) = {Tc20[-1]:.4f}")
Tc20 = np.array(Tc20)
print(f"\nTc_CNN(L=16) = {Tc_mean:.4f} +/- {Tc_std:.4f}")
print(f"Tc_CNN(L=20) = {Tc20.mean():.4f} +/- {Tc20.std(ddof=1):.4f}")

# Pendiente de la curva en el cruce: ¿es más abrupta para L = 20?
def pendiente(res):
    t, p = res["T"].to_numpy(), res["prob_media"].to_numpy()
    i = np.where((p[:-1] <= 0.5) & (p[1:] >= 0.5))[0][0]
    return (p[i+1] - p[i]) / (t[i+1] - t[i])
print(f"pendiente dP/dT en el cruce:  L=16 -> {pendiente(resumen16):.2f}   L=20 -> {pendiente(resumen20):.2f}")
````

Las dos estimaciones son compatibles dentro de su dispersión. Lo que sí cambia claramente es la **pendiente** de la curva en el cruce: pasa de 3.8 a 5.0 por unidad de temperatura. Para una red más grande ($L=20$) se espera que la transición se vea **más abrupta** (más cercana al límite termodinámico, donde la transición de fase es discontinua), mientras que en $L=16$ el efecto de tamaño finito la suaviza un poco. Aun así, la misma CNN — sin ver un solo ejemplo de $L=20$ durante el entrenamiento — debería ubicar el cruce cerca de $T_c$.

````python
# ===== 1. Kernels de la primera capa =====
conv1 = model.layers[0]
K, bK = conv1.get_weights()           # K: (3, 3, 1, 8)
K = K[:, :, 0, :]

k_vecinos = np.array([[0, 1, 0], [1, 0, 1], [0, 1, 0]], float)   # acoplamiento a primeros vecinos
k_promedio = np.ones((3, 3)) / 9                                  # promedio local ~ magnetización local
k_vecinos_centro = np.array([[0, 1, 0], [1, 4, 1], [0, 1, 0]], float)

def coseno(a, b):
    return float((a * b).sum() / (np.linalg.norm(a) * np.linalg.norm(b)))

fig, axes = plt.subplots(2, 6, figsize=(13, 4.6))
vmax = np.abs(K).max()
for k in range(8):
    ax = axes.flat[k]
    ax.imshow(K[:, :, k], cmap="RdBu_r", vmin=-vmax, vmax=vmax)
    ax.set_title(f"filtro {k}\nsesgo={bK[k]:+.2f}", fontsize=9)
    ax.set_xticks([]); ax.set_yticks([])
for ax, (kk, nombre) in zip(axes.flat[8:], [(k_vecinos, "vecinos"), (k_promedio, "promedio"),
                                            (k_vecinos_centro, "centro+vecinos")]):
    ax.imshow(kk, cmap="RdBu_r", vmin=-kk.max(), vmax=kk.max()); ax.set_title(f"referencia:\n{nombre}", fontsize=9)
    ax.set_xticks([]); ax.set_yticks([])
axes.flat[-1].axis("off")
plt.suptitle("Kernels aprendidos (rojo +, azul −) y kernels físicos de referencia"); plt.tight_layout(); plt.show()

tabla_k = pd.DataFrame({
    "suma pesos": K.sum(axis=(0, 1)),
    "peso centro": K[1, 1, :],
    "cos(vecinos)": [coseno(K[:, :, k], k_vecinos) for k in range(8)],
    "cos(promedio)": [coseno(K[:, :, k], k_promedio) for k in range(8)],
    "sesgo": bK,
}).round(3)
tabla_k.index.name = "filtro"
print(tabla_k)
print("\n|cos| ≈ 1 = mismo patrón (el signo solo indica si detecta dominios + o −)")
````

````python
# ===== 3. Sensibilidad a voltear un solo espín =====
def sensibilidad(x):
    """S(x) = (1/L^2) sum_i |P(x con s_i volteado) - P(x)| para una configuración x de forma (L, L, 1)."""
    Lx = x.shape[0]
    volteadas = np.repeat(x[None], Lx * Lx, axis=0)
    ii, jj = np.divmod(np.arange(Lx * Lx), Lx)
    volteadas[np.arange(Lx * Lx), ii, jj, 0] *= -1
    p0 = model.predict(x[None], verbose=0)[0, 0]
    p = model.predict(volteadas, verbose=0, batch_size=512)[:, 0]
    return np.abs(p - p0).mean(), p0

rng_s = np.random.default_rng(0)
n_por_T = 20
filas = []
for temp in np.unique(T_tc):
    idxs = rng_s.choice(np.where(T_tc == temp)[0], n_por_T, replace=False)
    for idx in idxs:
        S, p0 = sensibilidad(X_tc[idx])
        filas.append((temp, S, p0))
sens = pd.DataFrame(filas, columns=["T", "S", "P"])
sens_T = sens.groupby("T").agg(S_media=("S", "mean"), S_std=("S", "std"), P_media=("P", "mean"))

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
ax[0].errorbar(sens_T.index, sens_T["S_media"], yerr=sens_T["S_std"] / np.sqrt(n_por_T), fmt="o-", capsize=3)
ax[0].axvline(Tc_onsager, ls=":", color="k", label="Tc Onsager"); ax[0].set_yscale("log")
ax[0].set_xlabel("T"); ax[0].set_ylabel("S(x) promedio"); ax[0].legend(); ax[0].grid(alpha=.3)
ax[0].set_title("Sensibilidad a voltear un espín")
ax[1].scatter(sens["P"], sens["S"], s=6, c=sens["T"], cmap="coolwarm")
pp = np.linspace(0, 1, 100); ax[1].plot(pp, pp * (1 - pp) * sens["S"].max() / 0.25, "k--", label="∝ P(1−P)")
ax[1].set_xlabel("P(desordenada | x)"); ax[1].set_ylabel("S(x)"); ax[1].legend(); ax[1].grid(alpha=.3)
ax[1].set_title("S frente a la probabilidad de la configuración")
plt.tight_layout(); plt.show()

print(f"T con máxima sensibilidad: {sens_T['S_media'].idxmax()}   (Tc Onsager = {Tc_onsager:.3f})")
print(f"S lejos (T=1.6): {sens_T.loc[1.6, 'S_media']:.2e}   S cerca de Tc (max): {sens_T['S_media'].max():.2e}"
      f"   S lejos (T=3.2): {sens_T.loc[3.2, 'S_media']:.2e}")
````

````python
# ===== 4. ¿El logit final es una función de la magnetización? =====
def logits(modelo, Xs):
    """z = W·h + b de la última capa (antes de la sigmoide)."""
    penultima = keras.Model(inputs=modelo.inputs, outputs=modelo.layers[-2].output)
    h = penultima.predict(Xs, verbose=0)
    W, b = modelo.layers[-1].get_weights()
    return (h @ W + b).ravel()

# Misma arquitectura con pesos iniciales aleatorios (sin entrenar)
keras.utils.set_random_seed(42)
modelo_inicial = keras.models.clone_model(model)
modelo_inicial.build((None, None, None, 1))

m_tc = datos_tc["m"]                                  # |m| por configuración
m_firmado = X_tc.mean(axis=(1, 2, 3))                 # m con signo
z_ini = logits(modelo_inicial, X_tc)
z_fin = logits(model, X_tc)

fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))
for a, z, tit in [(ax[0], z_ini, "antes de entrenar"), (ax[1], z_fin, "después de entrenar")]:
    sc_ = a.scatter(m_tc, z, s=3, c=T_tc, cmap="coolwarm")
    a.set_xlabel("|m|"); a.set_ylabel("logit z"); a.set_title(f"logit vs |m| — {tit}"); a.grid(alpha=.3)
ax[1].axhline(0, ls="--", color="k")
plt.colorbar(sc_, ax=ax[1], label="T")
ax[2].scatter(m_firmado, z_fin, s=3, c=T_tc, cmap="coolwarm")
ax[2].set_xlabel("m (con signo)"); ax[2].set_ylabel("logit z"); ax[2].set_title("después de entrenar, vs m con signo")
ax[2].axhline(0, ls="--", color="k"); ax[2].grid(alpha=.3)
plt.tight_layout(); plt.show()

from scipy.stats import spearmanr
print(f"Correlación de Spearman logit vs |m|:  antes = {spearmanr(m_tc, z_ini)[0]:+.3f}   después = {spearmanr(m_tc, z_fin)[0]:+.3f}")
print(f"Correlación de Spearman logit vs m (con signo), después = {spearmanr(m_firmado, z_fin)[0]:+.3f}")

# ¿Cuánto del logit explica |m| solo? Ajuste polinomial z ≈ f(|m|) de grado 3
coef = np.polyfit(m_tc, z_fin, 3)
r2 = 1 - np.var(z_fin - np.polyval(coef, m_tc)) / np.var(z_fin)
print(f"R² de z ≈ polinomio(|m|) grado 3: {r2:.3f}")

# |m| en el que z = 0 (frontera de decisión expresada como magnetización)
mm = np.linspace(0, 1, 1001); m_frontera = mm[np.argmin(np.abs(np.polyval(coef, mm)))]
print(f"La frontera z = 0 corresponde aproximadamente a |m| ≈ {m_frontera:.2f}")

# Simetría s -> -s: ¿la red da lo mismo para x y -x?
p_x = model.predict(X_tc, verbose=0).ravel(); p_menos_x = model.predict(-X_tc, verbose=0).ravel()
print(f"max |P(x) - P(-x)| = {np.abs(p_x - p_menos_x).max():.3f}   media = {np.abs(p_x - p_menos_x).mean():.4f}")
````

### Respuestas — Interpretación de la red

**1. Kernels de la primera capa.** Los 8 filtros se agrupan en dos tipos:

- **Filtros 0, 1, 3 y 7: detectores de magnetización local.** Todos sus pesos tienen el mismo signo, la suma es $\approx \pm 1.6$ y el coseno con el kernel de promedio local es $|\cos| \approx 0.8$–$0.9$. Los filtros 1 y 7 (positivos) se activan en dominios de espines $+1$, y los filtros 0 y 3 (negativos) en dominios de $-1$. Con la ReLU, cada pareja funciona como un detector de "hay un dominio de este signo aquí". Juntas reconstruyen algo parecido a $|m|$ local.
- **Filtros 2, 4, 5 y 6: casi sin uso.** Sus pesos suman ≈ 0 y no se parecen a ningún kernel físico ($|\cos| < 0.2$). Tienen mezclas de signos que responden a bordes entre dominios o a ruido. Su activación media casi no cambia con $T$.

Ningún filtro se parece al kernel de "vecinos" puro ($\cos \le 0.82$, siempre menor que con el de promedio). La red no reconstruyó la energía del Hamiltoniano $\sum s_i s_j$: encontró el **parámetro de orden** (magnetización local), que es la forma más directa de separar las fases.

**2. Mapas de activación.** Coinciden con lo que uno diría a ojo:

- **Configuración ordenada ($T=1.6$, casi toda $-1$):** los filtros 0 y 3 se encienden en toda la red ("todo es un gran dominio negro") y los demás quedan apagados.
- **Cerca de $T_c$:** los mismos filtros se encienden en regiones grandes, pero con huecos donde hay dominios del signo contrario.
- **Configuración desordenada ($T=3.2$):** ningún filtro domina. Los filtros 1 y 7 marcan las islas de $+1$, y los filtros "de bordes" (2, 4, 5, 6) dan señal dispersa en toda la red.

La gráfica de activación media vs $T$ lo resume: los filtros de magnetización bajan de forma marcada al cruzar $T_c$.

Detalle de arquitectura: `padding="same"` rellena con ceros, por eso las activaciones caen en el borde de la imagen. Esto no es físico, porque los datos tienen **frontera periódica**; un relleno circular respetaría mejor la física.

**3. Sensibilidad a voltear un espín.** $S(x)$ tiene un pico muy marcado justo en $T = 2.275 \approx T_c$, con $S \approx 5\times10^{-2}$. Lejos de la transición es 200 veces menor ($2.7\times 10^{-4}$ a $T=1.6$) y 2000 veces menor a $T=3.2$. La razón es doble:

- **Matemática:** $\partial P/\partial z = P(1-P)$ es máxima cuando $P = 0.5$, y la gráfica de $S$ contra $P$ sigue esa forma.
- **Física:** cerca de $T_c$ la longitud de correlación diverge y el sistema está en el límite entre fases. Una perturbación pequeña puede inclinar la configuración hacia un lado u otro; es el análogo de la divergencia de la susceptibilidad $\chi$ en el punto crítico. Lejos de $T_c$, un espín volteado no cambia nada: un dominio grande sigue siendo un dominio grande, y el ruido sigue siendo ruido.

**4. ¿El logit es una función de la magnetización?** Sí, en gran medida:

- **Antes de entrenar**, el logit casi no depende de $|m|$: correlación de Spearman +0.22, en una franja estrecha de −0.5 a −0.3.
- **Después de entrenar**, la correlación es **−0.97** y un polinomio cúbico en $|m|$ explica el **94.5 %** de la varianza del logit. La frontera $z=0$ corresponde a $|m| \approx 0.77$.
- **Contra $m$ con signo**, el logit dibuja una parábola simétrica (correlación −0.06): la red aprendió que $+m$ y $-m$ son la misma fase. Es decir, aprendió la **simetría $\mathbb{Z}_2$** del modelo de Ising sin que se le dijera. No es exacta: $|P(x) - P(-x)|$ vale 0.04 en promedio y llega a 0.45 cerca de $T_c$, porque los filtros "positivos" y "negativos" no son espejos perfectos.

Conclusión: la red **redescubrió el parámetro de orden** por su cuenta. El resultado no es solo un truco de la arquitectura, aunque la arquitectura ayuda: `GlobalAveragePooling2D` promedia sobre todo el espacio, que es justo la operación con la que se define $m$. El 5 % de varianza que $|m|$ no explica corresponde a información adicional sobre el tamaño y la forma de los dominios.
